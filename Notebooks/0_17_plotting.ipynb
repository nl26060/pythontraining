# 17. Plotting graphs using matplotlib

MatPlotLib is the mathematical plotting library, created as an open source version of the MatLab plotting functionality. 

It is fully featured, you can plot pretty much anything with it, and there are lots of examples on the internet. 

In this notebook, there are a few examples of the types of graphs you will plot. The code is explained. At the end I suggest that you copy some of the graphs and play around with the code.

```python
# imports numpy, gives it the nickname np
import numpy as np
# imports matplotlib, gives it the nickname plt
import matplotlib.pyplot as plt
```

## Simple line graph

The graph below plots 1D data, it's plotted on the x axis.

```python
# plots 1D data
plt.plot([1, 3, 7, 5])
# plots a y axis label
plt.ylabel('some numbers');
# N.B. the semi colon prevents jupyter from printing the output of the last command to the screen
```

## Line plot of 2D data

The first argument is the x data, the second is the y data.

```python
plt.plot([10, 12, 32, 45], [1, 25, 22, 16])
plt.ylabel('some numbers')
plt.xlabel('some different numbers');
```

## Using plot markers

Graph as above, but now we add in some style options `:or`.

The `:` gives us dotted lines

The `o` gives us circular markers

The `r` colours it red

```python
plt.plot([10, 12, 32, 45], [1, 25, 22, 16], ':or');
```

##  Another example of a simple plot

Here we change the colour to green but do not plot the line.

`plt.axis([0.9, 5.5, 0, 26])` sets the axes limits, the numbers are:

`[xmin, xmax, ymin, ymax]`

```python
plt.plot([1, 2, 3, 4,5], [1, 4, 9, 16,25], 'go')
# sets the axis ranges
plt.axis([0.5, 5.5, 0, 26]);
```

## More complex line plot

First lets look at what the `np.linspace()` function does

```python
x = np.linspace(0, 10, 20)
x
```

It gives you linearly spaced numbers from xmin to xmax. This is useful for plotting functions. 

`np.linspace(xmin, xmax, number_of_points_desired)`

Now, lets discuss the graph below. Instead of inputting the data directly, we assign some data to x and then calculate y inside the plot function, e.g. using `x**2` for y in the quadratic plot. 

Also, we have plotted more than one line series of data here. 

```python
x = np.linspace(0, 2, 100)  # Sample data using np.linspace

plt.plot(x, x, label='linear')  # plot x versus x
plt.plot(x, x**2, label='quadratic')  # plot x vs x squared
plt.plot(x, x**3, label='cubic') # plot x vs x**3
# add labels to the graph
plt.xlabel('x label')
plt.ylabel('y label')
plt.title("Simple Plot")
# make a legend using the labels above
plt.legend(); 
```

## More compex plot

Here's a different way of doing the same thing. There is more than one way to produce a graph in matplotlib. (You probably want to use the simpler method above). N.B. we've used some new numpy code here: `np.tan()`.

```python
# sets the size, and makes an axes object that contains the graph
fig, ax = plt.subplots(figsize=(5, 2.7))
x = np.linspace(0, 1)
# plots on the axes objeect
ax.plot(x, x, color='blue', linewidth=3, linestyle='--', label='blue data')
l, = ax.plot(x, np.tan(x), color='orange', linewidth=2, linestyle=':',  label='orange data')
plt.xlabel('x label')
plt.ylabel('y label')
plt.title("Simple-ish Plot")
plt.legend();
```

## Scatter

The plot below is an example of a scatter plot. N.B. we've used some new numpy code here: `np.sin()`.

```python
# make x a set of 30 numbers between 0 and 10
x = np.linspace(0, 10, 30)
# make y sin of x
y = np.sin(x)
# plot a scatter plot with stars as the marker
plt.scatter(x, y, marker='*')
plt.xlabel("x")
plt.ylabel("Sin(x)");
```

## More complex scatter plot

This code shows you several different marker styles. It loops over each marker style and then plots some random data using a new numpy command.

```python
# assign rng to a random number generator
# it will randomly give you a number between 0 and 1
rng = np.random.RandomState()
# loops over marker styles
for marker in ['o', '.', ',', 'x', '+', 'v', '^', '<', '>', 's', 'd']:
    plt.plot(rng.rand(5), # random x data - 5 random values between 0 and 1
             rng.rand(5), # random y data
             marker, # which marker to use
             label="marker='{0}'".format(marker)) # label our data with marker
plt.legend(numpoints=1) # uses labels to make legend
plt.xlim(0, 1.8); # set x limits so there's room for the legend
plt.title('Random data');
```

## Error bar plots

You've seen this before and it's included here for completeness.

In this example, I've added a line plot. Here we're imagining that the error bar plot shows some experimentally measured data and the line plot shows a theory meant to explain it. Comparing theory to experiment is something you will do a lot in your scientific career.

```python
# make some data
my_2D_array=np.array([[3.9, 1, 1.5], [4,4.5,4.7], [7.7, 7.5, 9], [12, 13, 12.6]])
# calculate standard deviation
stds=np.std(my_2D_array, axis=1)
# calculate means
means=np.mean(my_2D_array, axis=1)

# make x_data 
x_data = [1,2,3,4]
# make an error bar plot
plt.errorbar(x_data, # x axis data
             means, # y axis data
             yerr=stds, # errors
             fmt="ok") # a different way to set the style

new_x_data = np.linspace(0, 4, 100)
# make y data from the theory
y_data = [x**2 for x in new_x_data]
# plot the theoretical predictions
plt.plot(new_x_data,y_data, '-')
# a different way to set up the legend
# note that you can use LaTeX maths in graphs by using the dollar signs
plt.legend(['theoretical model: $y=x^2$', 'experimental data'])
plt.title('Does the theory fit the data?')
plt.xlabel('x label')
plt.ylabel('y label');
```

## Histograms

The plot below generates a histogram. You can change the number of bins from 15 to another number (try it out!)

`plt.hist(x, number_of_bins)`

```python
# make some random data of mean mu and standard deviation sigma
# this data should be Gaussian (by the central limit thereom)
mu, sigma = 115, 15
x = mu + sigma * np.random.randn(10000)
# the histogram of the data
n, bins, patches = plt.hist(x, 15)

plt.grid(True) # add grid lines to the plot

plt.xlabel('x label')
plt.ylabel('y label')
plt.title("Histogram");
```

## Bar charts 

Below is an example of a bar chart with error bars. 

```python
# make x
x = [1, 2, 3, 4]
# assign means
means = [10, 12, 20, 22]
# assign standard deviations
stds = [2.3, 1.9, 0.5, 0.3]
# we're setting up an axis object here to attach the plots to
fig, ax = plt.subplots()
# make a bar chart
ax.bar(x, # x data
       means, # y data
       yerr=stds, # error bar
       align='center', # where to align the bar with the x value
       alpha=0.5, # darkness of the colour
       ecolor='black', # error bar colour
       capsize=10) # size of error bar caps
ax.set_ylabel('y label (units)') 
ax.set_xticks(x) 
# where to set the ticks
ax.set_xticklabels(['control', 'a', 'b', 'c'])
# what to label the ticks
ax.set_xlabel('x label (different units)')
ax.set_title('Awesome title here')
# make a grid
ax.yaxis.grid(True)

# Save the figure and show
plt.tight_layout()
# saves the plot
plt.savefig('bar_plot_with_error_bars.png')
```

N.B. If the error bars overlap, there is not a statistically significant difference between the data.
