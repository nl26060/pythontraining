Note that we are now using a **nested dictionary** and our function inputs two variables and returns two variables.

```python
def periodictable_all(elementname, property):
    
    ptable_extended = {
    'H': {'Atomic Mass': 1.00784, 'No. Protons': 1, 'Melting Point': -259.1, 'Boiling Point': -252.2},
    'He': {'Atomic Mass': 4.002602, 'No. Protons': 2, 'Melting Point': -272.2, 'Boiling Point': -268.9}
}
    ### continue for more elements

    element_details=ptable_extended[elementname]
    element_property=element_details[property]
    
    return element_details, element_property
```

Ideally, we want an easy way to create multiple **objects** which contain multiple variables. This can be achieved by using a **class**.

```python
class element_func():
    def __init__(self, name):
       self.name = name   

    def printer(self):
        print("This elements symbol is "+self.name)

### now create the object and call the function

Helium = element_func("He")

#
```
