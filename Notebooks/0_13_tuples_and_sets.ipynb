## Tuples

Lists and dictionaries can be useful when we need to manipulate data, however, sometimes we want the data to be **immutable** meaning that it cannot be change - effectively it is defined. In this case we want to use a tuple, which is defined by `()`

```python
a_tuple = ('x', 'y', 'z')
a_tuple
```

The tuple is also indexed, so we can extract a given value:

```python
a_tuple[0]
```

Tuples can also be homogenous or heterogenous depending on the data it contains.

```python
b_tuple = (10, -5, 7, 3)
c_tuple = ('Hello', 42, 3.14, True)

print(b_tuple)
print(c_tuple)
```

As stated tuples are **immutable**, try changing the following index in `c_tuple` to `3.14159`

```python
c_tuple[2] = 3.14159  
```

In a similar way you cannot append or delete items from a tuple, therefore if you want to take a tuple and add data to it you need to make a brand new tuple. In the following code we will start with a tuple, but change it by overwriting it with a new tuple

```python
d_tuple = ('kinetics', 'dynamics', 'thermodynamics')
print(f"d_tuple is {d_tuple}")
b_list = [5, -4, 3, 8]

d_tuple = tuple(b_list)
print(f"d_tuple is now {d_tuple}")
```

### Sets

Sets are the final form of built-in data types in Python (alongside lists, dictionaries and tuples) and are initiated with `{}`, they should not be confused with dictionaries which have **keys:values**. Sets are defined as **(1)** unordered, **(2)** mutable (elements can be added and removed, but not editted), **(3)** unindexed and **(4)** unique data storage types. All these propeties will now be displayed in the following blocks of code. Comment which block is displaying what from **(1)**, **(2)**, **(3)** and **(4)**.

```python
# 
my_set1 = {'a', 'b', 'c', 'd'}
my_set1[2]
```

```python
# 
my_set2 = {7, -1, 9, 2}
my_set2 
```

```python
# 
my_set3 = {'a', 'b', 'c', 'd'}
my_set3.add('e')
my_set3.discard('a')
my_set3
```

```python
# 
my_set4 = {1, 2, 2, 2, 2, 3, 4, 4}
my_set4
```

Unlike tuples sets have more functionality and operations can be performed on them. In each case describe what the operation is doing to the set with a comment.

```python
set_a = {45, 7, -5, 10}
set_b = {'yellow', 'purple', 'red', 'green', 'blue'}

list_a = ['orange', 'red', 'green', 'blue']
```

```python
# 
set_c = set_a.union(set_b)
set_c
```

```python
# 
set_d = set_a.union(list_a)
set_d
```

```python
#  
set_e = set_b.intersection(list_a)
set_e
```

```python
# 
set_f = set_b.difference(list_a)
set_f
```

```python
# 
set_g = set_b.symmetric_difference(list_a)
set_g
```
