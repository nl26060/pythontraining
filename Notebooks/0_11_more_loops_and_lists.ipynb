## Making a list of pairs

You can put any type of python variable into a list, even other lists! This is very useful. 

The code below will make a list of pairs from the two lists we declared at the top of this document.

```python
list_of_pairs=[]
for i in range(len(list_a)):
    list_of_pairs.append([list_a[i], list_b[i]])
    
list_of_pairs
```

There's some new stuff in this example. 

We tell python that something is a list by putting it into square brackets, `[ ]`. For a list of lists we have more than one set of brackets, e.g.:

`[ [ first list ] , [ second list ], [ third list ] ]`

You can pull out elements of the list of lists exactly as you can pull out elements of a list.
