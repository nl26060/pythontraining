# Coding a Game in Python

In this final exercise we will practise many of the skills learnt in Python over this course by coding a game in Python - you can do this task indvidually or in groups of up to three.

Your task is to create a game with the following premise:

1. You are in a reaction vessel of any shape or dimension you chose
2. The reaction vessel contains a given amount of reactive molecules, which move randomly
3. The player is a catalyst and must move in the vessel to convert the reactive molecules to safe molecules

Therefore, you must set a game up with these properties and it must be playable, ideally through the command line using a python script, noting you will need to use the `input()` command so the player can input specifications for the game and actually play it!

This is a difficult challenge, it is recommended not to use AI and try to code this yourself first e.g. how can you implement what you want in a for loop with the appropriate restrictions. As you grow in confidence with this challenge you can also increase the complexity of the game:

1. What if there are two types of molecules?
2. What if there is a set amount of turns?
3. What if you can power up the catalyst (double movement, molecules stay still etc.)?
4. What if you have to avoid a catalyst deactivator molecule?
5. What if the reaction vessel is stirred and everthing randomly moves?

As mentioned this task should push you to practise setting boundaries and writing complex code, for example you might want to write a for loop and then convert this to a function etc. to clean as you go.
