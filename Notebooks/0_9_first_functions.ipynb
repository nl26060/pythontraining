* We signal its execution (we click 'go!') by the use of brackets `()`.

* The `print()` function, like many others, accepts **arguments** - extra stuff inside the brackets that gets fed into the function to tune what it does.

* The `print` function also works *without* any extra arguments. However, we still need the `()` to signal to the interpreter that we want the result of *calling* the function and not just the function itself.

```python
print("This line of text is separated from another line of text by two blank lines, each blank line printed by a different technique.")

print("\n")

print()

print("Here's the second line of text!")

# if we don't include () we just get the function.
print
```

A function is simply a smaller python program that you write and then reference. Let's create a simple function **definition** that will print two things and then **call** it ourselves.

Hit `Shift+Enter` on the code blocks below.

```python
# This function is called print_two and takes two arguments
def print_two(a, b):
    """ prints two arguments
    a: first argument
    b: second argument
    """
    print(f"First argument in print_two function definition is: {a}")
    print(f"Second argument in print_two function definition is: {b}")
    print("\n")
    return

# Call the functions
print_two("Hello again, ","world!")
```

```python
# This function, print_none, takes no arguments
def print_none():
    """a broken print function!"""
    print("I got nothin.")
    return

print_none()
```

*Most* python functions, and all the ones you will write in this course, take a fixed number of arguments* - if you pass the wrong number, then the code will throw an error. For example, if you wrote:

`print_none("Please, please print me (print me!)")`

You will get the error :

`TypeError: print_none() takes 0 positional arguments but 1 was given`

which tells you that the function `print_none()` takes no arguments, but you gave it one argument (the string `"Please, please print me (print me!)"`), so this is an error as you have the wrong type of input. 

*The astute among you will realise that the built-in `print` function is one of the special functions that can take multiple numbers of arguments.

Print functions like we have looked at so far are actually a special type of function that do not return a value. These are sometimes called *procedures*.
Most functions act like mathematical functions and do something to their arguments before *returning* the resulting value. Consider the following function to add one to the number you've passed it:

```python
def add_one(input_value):
    """adds one to input value and returns that
    """
    return input_value + 1

x = 5
print(f"x is {x}")
# Note that you can safely store the result of a function in one of the passed in variables!
x = add_one(x)
print(f"x is now {x}")
# Also note that the print function can take any valid python statement as an inline parameter
print(f"and one more than that is {add_one(x)}")
```

Whilst some comments are provided in the code block above, **do not worry if this doesn't make any sense yet**! Here are some things to note right away:

* We use `def` to **define** a function.
* The name of the function (whatever you like) follows the `def` keyword.
* Brackets follow the function name.
* If your function needs some input to work properly, these can be included as **arguments** inside the brackets.
* All opening lines of a function definition (i.e. the line that has `def` on it) ends with a colon `:`.
* **Everything inside** a function is **indented** in order to signify what is in the function defintion and what is not.
* You *may* use block comments (""") to describe what your function does immediately after the definition. You do not have to do this, but should for complex or important functions, or ones you expect other people to use - these are called **docstrings**.
* If a function will **return** something, you use the *return* keyword with the value to return. This value comes out of the function so you can use it. For example, in the code above, x was set to 6 after calling the function, since the function returned the value 6 (5+1) and x was assigned to the output of the function

### Why bother with functions?

Throughout this introductory course, you have been asked to fix some bugs in your code.

The odd typo here, the missing bracket there... once you get to grips with the language, bugs are relatively easy to spot in code that's less than, say, 20 lines long.

Manually scanning for code bugs *will not* work when you get to more complex scientific code that can be thousands, or even *millions* of lines long!

Functions enable you to create little 'bubbles' of self-contained code. You can pass stuff in and get new stuff out.

The useful thing is that, if your program has a bug, functions make it much easier to track the bug and fix it!

The job of rewriting code to introduce functions rather than linear code is known as *refactoring*
