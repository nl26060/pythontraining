## SMILES and molecules as objects

If you remember in the dictionaries notebook we used a very simple method of inputting a chemical formula to allow us to use Python to calculate the molecular weight. There are nice methods or inputting chemical formulae into computers called things like SMILES, SMIRKS and SELFIES. These are strings which **encode** chemical formulae. Take a look at some of them in the cell below.

```python
df[['smiles']]
```

They might look like someone has leant on the keyboard, but you can learn to read them... (ish). The code below prints out the 5$^{\mathrm{th}}$ smiles string in the list:

```python
x=df['smiles'][4]
x
```

To read it:

there is a carbon ring, we know this as there is a 1 after the first carbon and a 1 at the end, indicating that the last carbon is linked to the first (*i.e.* the two carbons with 1 after them are joined). We know that there are 4 carbons and 1 sulfur in the ring. As they are lower case letters, we know that the ring is aromatic. Hydrogens are not included as it is assumed a chemist would be able to work out where they go from this information!

No one expects you to be able to read SMILES strings! What we can do is give the SMILES string to a computer program to read it. In this case we use a function from `rdkit`'s `Chem` function called `MolFromSmiles`, as below:

```python
rdkit.Chem.MolFromSmiles(x)
```

And we can see that this molecule is thiophene (and if you did not know that, you could have used the code below to read it from the entry in the DataFrame):

```python
df['Compound ID'][4]
```

This is also the first ever **object** that you have come across. Like real world objects it has **properties** and can do things, these are called **methods**. The output of `rdkit.Chem.MolFromSmiles(x)` is not a picture of a formula (although that is what is printed to the screen), it is an object. If you have ever played a computer game, you are familiar with objects, most objects you interact with in the game (blocks, guns, swords, spaceships, NPCs, angry birds, etc).

In the cell below you will define an object which you will call `thiophene_object`

```python
thiophene_object = rdkit.Chem.MolFromSmiles(x)
```

To see the structure, we print it to screen by putting it in a cell:

```python
thiophene_object
```

Now we use one of the built in **methods** called `GetNumAtoms`. As the method belongs to the object, we call it as `object.method()`, it is like a function, in that you can put specifications into the brackets, but you will not do that here:

```python
thiophene_object.GetNumAtoms()
```

This number does not seem right, trying another method, `GetNumHeavyAtoms`, counts the number of non hydrogen atoms:

```python
thiophene_object.GetNumHeavyAtoms()
```

So, in the `GetNumAtoms` the hydrogen atoms have not been counted:

```python
thiophene_object.GetNumAtoms() == thiophene_object.GetNumHeavyAtoms()
```


If you remember, the SMILES strings didn not have hydrogens in them. So we could use chemical knowledge to count the missing hydrogens...

or you can use `rdkit` to do it for you.

To do this you will use a rdkit method: `rdkit.Chem.AddHs` which will add the Hs to the structure. Call this method on the thiophene_object, by putting it in the `( )`, and then assigning this to a new object (`thiophene_object_H`): 

```python
thiophene_object_H = rdkit.Chem.AddHs(thiophene_object)
thiophene_object_H
```

Now the printed molecular structure has hydrogens on it. And you can check this by using similar code to above to count the number of atoms and the number of heavy atoms.

```python
thiophene_object_H.GetNumAtoms()
```

```python
thiophene_object_H.GetNumHeavyAtoms()
```

```python
thiophene_object_H.GetNumAtoms() == thiophene_object_H.GetNumHeavyAtoms()
```

It is now correct. You can count the number of hydrogens as well by doing some maths:

```python
thiophene_object_H.GetNumAtoms() - thiophene_object_H.GetNumHeavyAtoms()
```

## Investigating the dataset

Lets start by having a look at a few more molecules. 

You can put almost anything into a Python list, including objects. So, the code below makes a list of objects created from the first 12 molecules in the DataFrame. You do not need to understand the following **list comprehension** line of code, but it is worth looking at for a moment.

`df['smiles'][:12]` grabs the smiles column from the DataFrame, and just like in lists and arrays, we can use slices `:` to take parts of list. This takes the first 12 smiles strings (*note* that this logic can be applied in reverse with `df[:12]['smiles']`).

`rdkit.Chem.MolFromSmiles(x) for x` gives us the objects of the `MolFromSmiles` function called on each `x`. Where `x` is an element in the `df[:12]['smiles']` series. 

Finally as all of this code is in `[]`, this gives us a list of molecule objects.

```python
molecule_list = [rdkit.Chem.MolFromSmiles(x) for x in df['smiles'][:12] ]
molecule_list
```

The output of `molecule_list` is a list of objects (e.g. `<rdkit.Chem.rdchem.Mol at 0x2110b84a650>`) which tells us there is an rdkit Mol object and the weird looking number tells us exactly where on the computer it is stored (you will never need to know this!)

You can address this list as you can all other lists, for example here is the 9th object in the list:

```python
molecule_list[8]
```

But the really cool thing you can do with this is make a nice plot of the whole list, to do this you use `rdkit.Chem.Draw.MolsToGridImage` as below:

```python
rdkit.Chem.Draw.MolsToGridImage(molecule_list)
```

There are other descriptors out there (*e.g.* Mordred), but you may have noticed that this takes a little while to calculate, because the descriptors are information rich, therefore for very large databases even these descriptors become computationally costly, and we may need simpler representations. Therefore it is possible to featurise a molecule just based solely on its structure. The simplest approach is to use a **Molecular Fingerprint**

A molecular fingerprint is a binary encoding of a molecule where the 0 and 1 represent some form of property. There are many different types of fingerprints and the details on how each one is constructed will not be discussed here. But RDKit does allow us to construct some molecular fingerprints, let's use the Morgan Fingerprint on the thiophene molecule. 

```python
from rdkit.Chem import rdFingerprintGenerator
```

```python
fpgen = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=2048)

thiophene_fp =  fpgen.GetFingerprintAsNumPy(thiophene_object)
thiophene_fp
```

We have a NumPy array (a vector) of 0 and 1's which provide a quick and easy coding for the molecules to represent its properties. Create a DataFrame called `df_fp` to get the fingerprints of every molecules in the `delaney.csv`

```python
df_fp
```

You will notice this is much quicker, but it is now more difficult to interpret the information in a chemically meaniningful way. There are also many types of fingerprints available (lots in RDKit with control on parameters). It is a good idea to investigate a few in projects!

These molecular fingerprints can be directly read into a ML algorithm. However once useful thing that can be done with them out the gate is to examine the "similarity" of molecules. The concept of similarity is a complex topic, one could view this is how close they are in **euclidean** "chemical space" with some of the RDKit descriptors. However, with binary data the **euclidean** distance does not have the same meaning. However mathematical forumla to quantiy similarity for this kind of data is commonly used. Again, we will not focus on the forumlation but a common approach is to use the **Tanimoto Similarity**. For example we will look at the similarity between Thiophene and Citral, first we now process the fingerprint as an object rather than a NumPy array. Perform a similar analysis to obtain `citral_fp_obj` and run the code to obtain the similarity. 

```python
thiophene_fp_obj =  fpgen.GetFingerprint(thiophene_object)
thiophene_fp_obj 
```

```python
#
```

```python
from rdkit import DataStructs

DataStructs.TanimotoSimilarity(thiophene_fp_obj,citral_fp_obj)
```
