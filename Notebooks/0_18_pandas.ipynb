# 18. Pandas

Previously you will have seen the package **NumPy**, now you will be looking at a python package called **Pandas**

Pandas is a useful tool for data science and analysis, as it is similar to Excel in many ways. 

**What will you use Pandas for?**

Primarily in this notebook, you'll use pandas to read and analyze experimental data in python, especially from *Excel files* (.xls or .xlsx) or *comma-separated value* (.csv) files.

**Why use pandas instead of Excel?**

When your data is stored in a .csv or .xls/.xlsx file, you can set it to read-only to prevent accidental changes. However, when using Excel, it is easy to unintentionally alter values, even by simply pressing a key by mistake. Much scientific data has been corrupted in this way. In fact, there have been cases where gene names were accidentally changed to dates in Excel without anyone noticing, leading to significant issues with the data.

By using pandas, you avoid opening the raw data file directly. Instead, you read it into Python, manipulate it there,  and keep a record of any changes in a Jupyter notebook. This reduces the risk of unintentionally damaging your data.

**How will we use pandas?**

While Pandas offers many advanced features, this course will primarily cover the basics.

Pandas uses something called a **DataFrame**, which looks and acts a lot like a spreadsheet of data. 

There are many ways to set up DataFrame, one easy way is to use **dictionaries**, which were discussed in an earlier notebook. 

To recap dictionaries have **keys** which have assigned **values**. In a dictionary the keys and values can be almost any data type. 

In the example below, you will create a dictionary called `data` with the keys as strings (specifically the names of fruit) and the values as a list of numbers.

```python
data = {
    'Apples': [3, 2, 0, 1], 
    'Oranges': [0, 3, 7, 2]
}

data
```

Now, we are going to load this data into a DataFrame. 

You do this by using the function: 
`pd.DataFrame()`

In the example below you are assigning the DataFrame to a variable called `df`:

```python
df = pd.DataFrame(data)

df
```

You will see that the *keys* from the `data` dictionary have become the column headings in the `df` DataFrame.

Additionally, the *values* from the dictionary are the entries in each row of `df`.

The **Index** of the `df` DataFrame is automatically generated when it is created, these are the numbers 0-3 on the left-hand side. 

But you could also create your own index when you initialize the DataFrame. For example you could use names as the index:

```python
df = pd.DataFrame(data, index=['Amy', 'Bob', 'Claire', 'David'])

df
```

Similar to how you explored **slicing** in NumPy arraies, slicing can also be achieved in pandas with `iloc` (**i**nteger-based **loc**ation) and `loc` (label-based **loc**ation).

*e.g.* to get the first row (remember you count from 0):

```python
df.iloc[0]
```

The first row can also be obtained through its label:

```python
df.loc['Amy']
```

to get the first column, you can first select all rows (`:`), followed by selecting column 1:

```python
df.iloc[:,0]
```

Again here it is easier to just select the column based on its name, for example:

```python
df['Apples']
```

You can simultanious extract both rows and columns, for example using `iloc` in combination with a *list* `[ ]`:

```python
df.iloc[[0,3],0]
```

Alternatively you can use `loc` in a similar way:

```python
df.loc[['Amy', 'David'],'Apples']
```

Alternatively the series can be directly turned into a list or array, for example using a NumPy **array**:

```python
np.array(df['Apples'])
```

To summarise using a single `[ ]` gives you the information as a **series** which can then be turned into a list or array.

But using double `[[ ]]` you can return the column as a DataFrame, for example:

```python
df[['Apples']]
```

Using the double `[[ ]]` you can also extract rows as a DataFrame, for example:

```python
df.iloc[[0,3]]
```

Again a combination can be used to extract select data, for example:

```python
df.iloc[[0,3],[0]]
```

Typically you will want to load in data from an external source, rather than type it yourself in a Jupyter Notebook. In this case we have a .csv file we want to analyse. 

We can read this file in using the `read_csv` function, which loads in a .csv file and assigns it to a new DataFrame, in this case called `metal_df`. 

(You will also find the function `read_xls` useful, as it reads in excel files - but you might need to specify the sheet).

In the cell below we take a look at this file. You can see that the first column indicates the sample number, followed by 3 additional columns referring to different metals.

```python
metal_df = pd.read_csv('metal.csv')

metal_df
```

This has added an extra column with a generic index. This isn't particularly useful, so we can instead use the `read_csv` function to set the *Sample* column as the *index* column (you can think about how `iloc` and `loc` will slice different rows now):

```python
metal_df = pd.read_csv('metal.csv', index_col=0)

metal_df
```

When you have large DataFrames, the function `.head` is useful, it gives you the top 5 entries so you can quickly see your data:

```python
metal_df.head()
```

A related function called `.tail` also exists, try it out by typing `metal_df.tail()` in the box below to see what it does:

This dataset contains the individual measurements of thermal expansion for 10 samples of a metal. The task was to measure the thermal expansion of each metal. Therefore, you will want to calculate the average of these measurements and the standard deviation.

There are multiple ways to achieve this outcome, here the means and standard deviations will be calculated separately for each species from the `metal_df[' ']` series followed by `np.mean` and `np.std`:

```python
# Calculate the average:
aluminum_mean = np.mean(metal_df['Aluminum'])
copper_mean = np.mean(metal_df['Copper'])
steel_mean = np.mean(metal_df['Steel'])

# print to screen - here uses a different print format to only print 3 decimal places
print("Aluminium: {:.3}".format(aluminum_mean))
print("Copper: {:.3}".format(copper_mean))
print("Steel: {:.3}".format(steel_mean))
```

In a similar way the standard deviation can be obtained:

```python
# Calculate the standard deviation:
aluminum_std = np.std(metal_df['Aluminum'])
copper_std = np.std(metal_df['Copper'])
steel_std= np.std(metal_df['Steel'])

# print to screen - here uses a different print format to only print 3 decimal places
print("Aluminium: {:.3}".format(aluminum_std))
print("Copper: {:.3}".format(copper_std))
print("Steel: {:.3}".format(steel_std))
```

Using the above variables you can now construct lists of the means and standard deviations:

```python
means = [aluminum_mean, copper_mean, steel_mean]
stds = [aluminum_std, copper_std, steel_std]
```

This data can then be used to construct a bar chart of the results:

```python
# Use the original DataFrame to extract the titles of the metals with list comprehension

materials = [x for x in metal_df]

# Build the plot

plt.bar(materials, means, yerr=stds, align='center', alpha=0.5, ecolor='black', capsize=10)

# Set the title and axis

plt.title('Coefficent of Thermal Expansion (CTE)')
plt.xlabel('Metal')
plt.ylabel('Coefficient of Thermal Expansion ($° C^{-1}$)')

# Set the design and save as a figure

plt.grid(True)
plt.tight_layout()
plt.savefig('bar_plot_with_error_bars.png')
plt.show()
```
