Just as a class provides organisation, we may want to make different modules (python scripts) which perform different tasks. Create a `circle.py` to create a module which calculates the perimeter and area of a circle when provided with a radius. Import this module and show it works in the code blocks below. 

```python
import circle
```

```python
first_circle = circle.geometry(3)
diameter = first_circle.diameter()
radius = first_circle.area()

print(f"diameter is {diameter} and area {radius}")
```

The circle and calc moldule clearly have different purposes, but they might also be related, they are both teaching us some basic mathematics, therefore we might want to make a `mathermatics` package. The first step in making a package is to make a directory called `mathematics` and copy your `calc.py` and `circle.py` into this directory, renaming them `my_calc.py` and `my_circle.py` as to not confuse the imported modules.

The key to making a package is to have a `__init__.py` script in the directory you have just made, you can view this script as a way of telling python that this directory is a package. First let's make an `__init__.py` file in mathematics, add a docstring to the top to descibe the package then save. Look at the below code is `mathematics` available, is `mathematics.my_calc` available? 

```python
import mathematics
```

```python
mathematics
```

```python
mathematics.my_calc
```

In this case we can import mathematics as the `__init__.py` has marked it as a package, however `my_calc` and `my_circle` are not attributes of mathematics. Effectively the `__init__.py` is just being marked as a package and nothing else. To use the modules we can import them directly, there are two syntax which allow you to access the module `import mathematics.my_calc` and `from mathematics import my_circle` try these two and comment the difference (before importing it is recommended to restart the kernel).

Effectively by importing modules in `__init__.py` a user does not have to dig around to figure out what each python script is called and they can directly access the functionality. The final thing we will do in our `__init__.py` file is add some variables which allow us to keep track of what is going on. Add the following lines of code, then run the cell below (restart the kernel):

`__version__ = 1.0.0`

`__author__ = your.name`

Note we keep seeing these `_` **dunder** some have specific meanings like the `__init__` to perform some specific function, however we can add our own as a coding convention to specify that this is a special metadata variable.

```python
import mathematics 

print(f"The package version is {mathematics.__version__}")
print(f"The package was made by {mathematics.__author__}")
```
