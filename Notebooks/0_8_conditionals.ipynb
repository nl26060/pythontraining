# 8. `if` This, Then...

A lot of what you have learned in this course has led up to this moment.

You will learn how to use the `if` statement.

This is Python's way of letting you design those possible forks in the road. How you determine where the program goes is determined by what you state should be done in the case that certain statements are `True`, `False`, or some combination of such outcomes.

`if` statements enable you to run code only when certain conditions are met.

Let's dive in with an example.

```python
# Define some variables, assigning integer values to each
people = 20
cats = 30
dogs = 15

# If the integer value of people is less than cats, run the indented print statement 
if people < cats:
    print("Too many cats! The world is doomed!")

#If the integer value of people is greater than cats, run the indented print statement     
if people > cats:
    print("Not many cats! The world is saved!")

#
if people < dogs:
    print("The world is overrun by dogs!")

#    
if people > dogs:
    print("We need more dogs!")

# Increment the value of the dogs variable by 5.
dogs += 5

#
if people >= dogs:
    print("People are greater than or equal to dogs in number.")

#
if people <= dogs:
    print("People are less than or equal to dogs in number.")

#    
if people == dogs:
    print("People are dogs.")
```

### Exercise 8b: `if` and `else`

When we make choices, it's rarely just about one scenario; A or B.

If this happens, then we do that... but what if **this** doesn't happen? Then what?

We need an `else`! 

We need to allow for **other scenarios** to play out to truly create forks in the road. Not just A or B, but A or B or C or D, and so on.

In the code blocks below, you will learn to apply `if` alongside its two partner terms: `else` and `elif` (meaning else if).

**NOTE - the code below introduces the new keywords `else` and `elif`...but it will not run right away. There are some bugs in this code that you will need to fix. Treat this part as an exercise in reinforcing what you know about the syntax of using the `if` statement.**

```python
people = 30
cars = 40
trucks = 15


if cars > people
    print("We should take the cars.")

elif cars < people:
    print("We should not take the cars.")
    
else:
    print("We can't decide.")


if trucks > cars:
print("That's too many trucks.")
    
elif trucks < cars:
    print("Maybe we could take the trucks.")
    
else:
    print("We still can't decide.")


if people > trucks:
    print("Alright, let's just take the trucks."
    
else:
    print("Fine, let's stay home then.")
```

### Exercise 8d - Making Decisions

Some of the earliest video games were little more than expert use of `if` statements. 

What elevated the simple `if` statement to something approaching a game, was combining:

1. User `input` with
2. `if` statements **nested** in other `if` statements.

The Python code below will help you learn how to craft programs that make decisions that are beyond a single **if this then that** step.

Click on the code block below and hit `Shift+Enter`.

Note: The code block below requires your input after you run it!

```python
print("You enter a dark room with two doors. Do you go through door #1, #2 or #3?")

door = input(">")

if door == "1":
    print("There's a gathering of famous dead scientists here. What do you do?")
    print("1. Go speak to Richard Feynman.")
    print("2. Go speak to Marie Curie.")
    print("3. No one.")

    scientist = input(">")

    if scientist == "1":
        print("Feynman trains you to be a stellar teacher. Good job!")
    elif scientist == "2":
        print("Curie imparts endless curiosity on you. Enjoy!")
    else:
        print(f"No conversation for you!")

elif door == "2":
    print("Welcome to the hall of delightful documentarians! Who do you wish to speak to?")
    print("1. David Attenborough.")
    print("2. Alice Roberts.")
    print("3. No one.")

    insanity = input(">")

    if insanity == "1" or insanity == "2":
        print("You are overcome with the inimitable joy of science communication. Happy storytelling!")
    else:
        print("Your quest for inspiration continues...")
        
elif door == "3":
    print("Welcome to the code mines! Who do you wish to speak to?")
    print("1. Charles Babbage.")
    print("2. Ada Lovelace.")
    print("3. No one.")

    coder = input(">")

    if coder == "1":
        print("Babbage shows you the plans for his famous difference engine. It would have been the world's first computer if he'd not gotten distracted and instead finished it! You vow to complete your projects!")
    elif coder == "2":
        print("The first computer programmer shows you her code. You find a bug.")
    else:
        print(f"No conversation for you!")

else:
    print("No door? No game play for you!")
```
